# Notebook 1 — Comprendre la descente de gradient sur une courbe 2D

## Objectifs

Dans ce notebook, nous allons apprendre à :

- distinguer une fonction, sa dérivée et son gradient ;
- comprendre la direction indiquée par la dérivée ;
- effectuer une étape de descente de gradient à la main ;
- programmer plusieurs étapes avec une boucle ;
- visualiser le trajet suivi jusqu'au minimum.

> Ici, la fonction possède une seule variable. Son graphe est donc une courbe dans un repère 2D : l'axe horizontal représente $x$ et l'axe vertical représente $f(x)$.

## 1. Principe mathématique

Nous étudions la fonction :

$$f(x)=(x-3)^2+1$$

Sa dérivée est :

$$f'(x)=2(x-3)$$

La dérivée indique la pente locale de la courbe :

- si $f'(x)>0$, la fonction monte lorsque $x$ augmente ;
- si $f'(x)<0$, la fonction descend lorsque $x$ augmente ;
- si $f'(x)=0$, la tangente est horizontale : nous avons un point critique.

La descente de gradient utilise la règle :

$$x_{nouveau}=x_{actuel}-\alpha f'(x_{actuel})$$

Le symbole $\alpha$ représente le **taux d'apprentissage**. Le signe moins nous fait avancer dans la direction opposée à la pente, donc vers la diminution de la fonction.

## 2. Importer les bibliothèques

```python
# NumPy permet de créer des tableaux numériques et une grille de valeurs de x.
import numpy as np

# Matplotlib permet de construire les graphiques de la fonction et du trajet.
import matplotlib.pyplot as plt

# Cette option rend les graphiques plus lisibles dans le notebook.
plt.style.use("seaborn-v0_8-whitegrid")
```

### Explication

- `numpy`, abrégé en `np`, servira à calculer plusieurs valeurs simultanément ;
- `matplotlib.pyplot`, abrégé en `plt`, servira à tracer la courbe ;
- le style choisi change seulement l'apparence, jamais les calculs.

## 3. Programmer la fonction et sa dérivée

```python
def fonction(x):
    """Calcule f(x) = (x - 3)² + 1."""
    # Python effectue d'abord x - 3, élève le résultat au carré, puis ajoute 1.
    return (x - 3) ** 2 + 1


def derivee(x):
    """Calcule la dérivée f'(x) = 2(x - 3)."""
    # Cette valeur mesure la pente de la courbe au point x.
    return 2 * (x - 3)


# Vérification au point x = 0.
x_test = 0.0
print(f"f({x_test}) = {fonction(x_test):.2f}")
print(f"f'({x_test}) = {derivee(x_test):.2f}")
```

Au point $x=0$, la fonction vaut $10$ et la dérivée vaut $-6$.

La pente négative signifie que la courbe descend lorsque nous nous déplaçons vers la droite. Pour réduire $f(x)$, il faut donc augmenter $x$. La formule de descente de gradient produira exactement ce mouvement.

## 4. Effectuer une seule étape de descente

Nous partons de $x=0$ avec un taux d'apprentissage $\alpha=0.1$. Avant de créer une boucle, décomposons une étape complète.

```python
# Point de départ choisi sur l'axe horizontal.
x_actuel = 0.0

# Le taux d'apprentissage contrôle la taille du déplacement.
taux = 0.1

# Calcul de la pente au point actuel.
pente = derivee(x_actuel)

# Application de la règle : nouveau = actuel - taux × pente.
x_nouveau = x_actuel - taux * pente

# Calcul des hauteurs avant et après le déplacement.
valeur_actuelle = fonction(x_actuel)
valeur_nouvelle = fonction(x_nouveau)

# Affichage détaillé pour vérifier chaque quantité.
print(f"Point actuel       : x = {x_actuel:.2f}")
print(f"Pente              : f'(x) = {pente:.2f}")
print(f"Déplacement        : {-taux * pente:.2f}")
print(f"Nouveau point      : x = {x_nouveau:.2f}")
print(f"Valeur avant       : f(x) = {valeur_actuelle:.2f}")
print(f"Valeur après       : f(x) = {valeur_nouvelle:.2f}")
```

### Calcul obtenu

La mise à jour est :

$$x_{nouveau}=0-0.1(-6)=0.6$$

La valeur de la fonction passe de $f(0)=10$ à $f(0.6)=6.76$. Une seule étape nous a donc rapprochés du minimum $x=3$.

## 5. Visualiser la pente et le premier déplacement

```python
# Création de 400 valeurs régulièrement espacées entre -2 et 7.
x_grille = np.linspace(-2, 7, 400)

# Calcul de la fonction pour les 400 valeurs grâce à NumPy.
y_grille = fonction(x_grille)

# Construction d'une petite zone autour de x_actuel pour tracer la tangente.
x_tangente = np.linspace(x_actuel - 1.2, x_actuel + 1.2, 100)

# Formule de la tangente : f(a) + f'(a)(x - a), avec a = x_actuel.
y_tangente = fonction(x_actuel) + derivee(x_actuel) * (x_tangente - x_actuel)

# Création de la figure.
fig, ax = plt.subplots(figsize=(10, 6))

# Tracé de la fonction complète.
ax.plot(x_grille, y_grille, label="f(x) = (x - 3)² + 1", linewidth=2)

# Tracé de la tangente en pointillés. Sa pente vaut f'(x_actuel).
ax.plot(x_tangente, y_tangente, "--", label="Tangente au point de départ")

# Le point rouge représente la position avant la mise à jour.
ax.scatter(x_actuel, valeur_actuelle, s=90, color="crimson", zorder=5, label="Avant")

# Le point vert représente la position après une étape.
ax.scatter(x_nouveau, valeur_nouvelle, s=90, color="green", zorder=5, label="Après une étape")

# La flèche montre le déplacement produit par la mise à jour.
ax.annotate(
    "",
    xy=(x_nouveau, valeur_nouvelle),
    xytext=(x_actuel, valeur_actuelle),
    arrowprops={"arrowstyle": "->", "color": "black", "lw": 2},
)

# Le minimum théorique est indiqué par une étoile.
ax.scatter(3, 1, marker="*", s=180, color="goldenrod", zorder=6, label="Minimum global")

# Ajout des informations nécessaires à la lecture du graphique.
ax.set_title("Une étape de descente de gradient")
ax.set_xlabel("Position x")
ax.set_ylabel("Valeur f(x)")
ax.legend()
plt.show()
```

### Comment lire ce graphique ?

La tangente descend de gauche à droite : sa pente est négative. Le gradient pointe donc vers la gauche, qui est ici la direction d'augmentation la plus rapide. Le signe moins de la descente de gradient nous fait partir vers la droite, donc vers des valeurs plus petites de la fonction.

## 6. Répéter la mise à jour avec une boucle

```python
def descente_gradient_1d(derivee_f, x_depart, taux, nombre_iterations):
    """Applique la descente de gradient à une fonction d'une variable."""

    # La position courante commence au point choisi par l'utilisateur.
    x = float(x_depart)

    # Cette liste conservera toutes les positions, y compris le départ.
    historique = [x]

    # La boucle répète exactement la même règle nombre_iterations fois.
    for _ in range(nombre_iterations):
        # Calcul de la pente à la position courante.
        gradient = derivee_f(x)

        # Déplacement dans la direction opposée au gradient.
        x = x - taux * gradient

        # Enregistrement de la nouvelle position pour la visualiser ensuite.
        historique.append(x)

    # Conversion en tableau NumPy pour faciliter les futurs calculs.
    return np.array(historique)
```

### Structure de la fonction

- `derivee_f` permet d'utiliser la fonction avec différentes dérivées ;
- `x_depart` fixe la position initiale ;
- `taux` contrôle la longueur de chaque pas ;
- `nombre_iterations` indique le nombre de mises à jour ;
- `historique` mémorise le trajet complet au lieu de garder seulement le résultat final.

```python
# Lancement de quinze étapes depuis x = 0 avec un taux de 0.1.
positions = descente_gradient_1d(
    derivee_f=derivee,
    x_depart=6.0,
    taux=0.1,
    nombre_iterations=15,
)

# Calcul de la hauteur de la fonction pour chaque position visitée.
valeurs = fonction(positions)

# Affichage des six premières étapes sous forme de tableau texte.
print(f"{'Itération':>9} | {'x':>10} | {'f(x)':>10} | {'gradient':>10}")
print("-" * 49)
for iteration in range(6):
    x = positions[iteration]
    print(f"{iteration:9d} | {x:10.6f} | {fonction(x):10.6f} | {derivee(x):10.6f}")

# Affichage du résultat après la dernière étape.
print(f"\nPosition finale : {positions[-1]:.6f}")
print(f"Valeur finale   : {valeurs[-1]:.6f}")
```

À chaque itération, $x$ se rapproche de $3$, $f(x)$ se rapproche de $1$ et la dérivée se rapproche de zéro. Le déplacement devient donc naturellement de plus en plus petit.

## 7. Visualiser le trajet complet et la convergence

```python
# Création de deux graphiques placés côte à côte.
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Premier graphique : positions successives sur la courbe.
axes[0].plot(x_grille, y_grille, linewidth=2, label="f(x)")
axes[0].plot(positions, valeurs, "o-", color="crimson", label="Trajet")
axes[0].scatter(3, 1, marker="*", s=180, color="goldenrod", zorder=5, label="Minimum")
axes[0].set_title("Trajet sur la fonction")
axes[0].set_xlabel("Position x")
axes[0].set_ylabel("Valeur f(x)")
axes[0].legend()

# Deuxième graphique : valeur de la fonction selon l'itération.
iterations = np.arange(len(valeurs))
axes[1].plot(iterations, valeurs, "o-", color="navy")
axes[1].axhline(1, linestyle="--", color="goldenrod", label="Valeur minimale = 1")
axes[1].set_title("Convergence de la valeur de la fonction")
axes[1].set_xlabel("Itération")
axes[1].set_ylabel("Valeur f(x)")
axes[1].legend()

# Ajustement automatique de l'espacement entre les graphiques.
plt.tight_layout()
plt.show()
```

## 8. Ce qu'il faut retenir

1. La dérivée mesure la pente locale.
2. Le gradient pointe vers l'augmentation de la fonction.
3. Le signe moins nous dirige vers la diminution.
4. Le taux d'apprentissage contrôle la taille du pas.
5. Le gradient diminue lorsque nous approchons du minimum.
6. Dans cet exemple convexe, l'unique minimum local est également le minimum global.
